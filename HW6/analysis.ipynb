**2. What do the earliest principal components (the first few "Eigenfaces") physically capture in the images? Why do later principal components appear increasingly like fine-grained pixel noise?**

The earliest principal components capture lighting and the general layout of the face. PC 1 captures how bright the face is compared to the background, PC 2 and PC 3 capture lighting patterns across the face, and the next few capture specific features like the mouth (PC 5), the eyes and nose (PC 6), and glasses (PC 7 and PC 8). Later principal components appear increasingly like fine-grained pixel noise because PCA orders components by how much variance they explain, so once the patterns shared by many photos are covered, the later components are built from small differences that only show up in one or two photos, like a slight head shift, a strand of hair, or camera noise.

**3. Why is it critical to fit the PCA transformer strictly on the training split rather than on the entire dataset prior to splitting? What negative impacts occur if PCA is fitted on the full dataset before training downstream classifiers?**

It is critical to fit the PCA transformer strictly on the training split because otherwise the test images help decide what the components look like, which leaks information about the test set into the model. If PCA is fitted on the full dataset before training downstream classifiers, the test accuracy looks better than it really is, the comparison between models becomes less trustworthy, and choices like how many components to keep end up being made using test data.

On PCA-reduced features, Logistic Regression had the highest accuracy (0.9625), compared to the Ensemble Classifier (0.9500) and the Kernel SVM (0.9375). Non-linear kernels and ensemble methods did not yield substantial accuracy gains, and both were slightly worse than Logistic Regression, with only 2 test images separating the best and worst model. Classical classifiers hit a performance ceiling on flattened pixel vectors or global PCA projections because these features don't keep track of which pixels are next to each other, so a small change in head position or expression changes many pixel values even for the same person, and PCA components capture the biggest changes across all the photos (mostly lighting) rather than the differences between people.
