How do the feature types in the UCI Heart Disease dataset (combining continuous physiological telemetry like chol and thalach with categorical/ordinal indicators like cp and thal) differ from previous datasets, and why is proper scaling critical for Support Vector Machines compared to Decision Trees?**

The feature types in the UCI Heart Disease dataset differ from previous datasets because while they might be categorized a certain way, the data itself tells a different story, or might have conflicting reasons to be categorized a certain way. For example, 'ca' represents a discrete integer value (0-3), which might indicate that it can be used categorically. However, based on its description which states that it enumerates the amount of major vessels colored by flourosopy, it could be used numerically as well. Overall, the feature types in this dataset are more superfluous and difficult to determine compared to previous datasets.

Proper scaling is critical for SVMs because SVMs measure distance between points. If for one variable these distances are large, but for another they are small, the one with larger distances/values will weigh more than the smaller one. But, if both are scaled, they can be weighed equally by the SVM. Decision Trees, on the other hand, are comparing one feature at a time to a threshold value, which means scaling would have no effect.

**2. Explain how hidden missing values (such as ? symbols) impact raw medical datasets, and justify your choice of imputation strategies for skewed numerical vitals versus categorical diagnostic fields.**

Hidden missing values like ? aren't recognized as missing. Pandas reads them as text, so isnull() doesn't count them, and the whole column is treated as text instead of numbers. That breaks the median, scaling and model training, or the ? gets encoded as if it were a real category. That’s why I checked for ? before checking for nulls. This dataset had none, but ca and thal had regular NaNs.

I filled the gaps instead of dropping rows, because dropping would throw away patients from an already small dataset. For skewed numeric vitals like cholesterol, the median is the better fill value, because a few extreme patients pull the mean away from a typical value but barely change the median. For categorical fields like thal, the most frequent value is the only sensible choice, since you can't average labels. I also used most-frequent for ca, which counts vessels from 0 to 3, because it always fills in a real count rather than a fraction.

**3. Contrast how a Support Vector Classifier (using an RBF kernel) constructs decision boundaries versus how a Decision Tree recursively partitions feature space. What are the interpretability tradeoffs between the two models?**

An SVC with an RBF kernel draws one smooth, curved boundary through feature space. It looks for the boundary with the widest possible margin between the two classes, using the distances between patients. C sets the tradeoff between a wide margin and few training mistakes, and gamma sets how tightly the curve bends around individual points. The final boundary depends on all the scaled features at once.

A Decision Tree splits the data one question at a time. At each node it picks the single feature and cutoff value (for example, “oldpeak ≤ 1.5?”) that best separate the two classes. Then it repeats the process on each side. This divides the feature space into rectangular regions with straight edges that run parallel to the feature axes. In my model, max_depth=4 and min_samples_leaf=5 limit how many times it can split, which keeps it from overfitting.

For interpretability, the tree is easy to follow. My plot_tree diagram in Task 8 shows which features and cutoffs lead to a “Disease” or “No Disease” prediction, and you can trace any patient's path through it like a flowchart. The RBF SVC is harder to comprehend. Its boundary is a curved surface built from distances between patients, so you can't easily say which simple rule caused a prediction. The tradeoff is that the SVC's smooth boundary often fits the data more accurately, while the tree gives up some of that flexibility in exchange for distinct rules that can be checked.

**4. Discuss how tuning hyperparameters such as regularization parameter C and kernel coefficient gamma in an SVC impacts the trade-off between bias, variance, and model generalization on unseen medical test data.**

C affects how strictly the SVC will fit the training data. If I set it to a low number, it will be less strict. This means I might underfit my data and increase bias since I am increasing the margin for error. If I increase my C value, my model becomes more strict. I run the risk of overfitting if the data is noisy, but am less biased.

My Gamma value controls the influence of each observation on observations around it. If I set it to a low value, each observation has more influence. This will increase the bias of my model and make it more generalized, but it will have less variance. A high Gamma value does the opposite. It lowers the individual influence of each observation, lowering the bias of my model, but making it more specialized, therefore prone to overfitting and variance.
